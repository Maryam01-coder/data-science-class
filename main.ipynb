```python
#30 DAYS DAILY STEPS TRACKER
np.random.seed(42) #reproducibility
days = np.arange(1, 31)
steps = np.random.randint(3456, 12034, size=30)

print("Days:", days)
print("Steps", steps)
```

```python
data = pd.DataFrame({
    "Days": days,
    "Steps": steps
})

data
```

```python
#basic stats
mean_step = np.mean(steps)
std_step = np.std(steps) #std= standard deviation

print("Average step:", mean_step)
print("Median step:", np.median(steps))
print("Minimun step:", np.min(steps))
print("Maximum step:", np.max(steps))
print("Standard deviation step:", std_step)
```

```python
#goal analysis for days of over 10000 steps
goal = np.where(steps > 10000)[0] + 1
print(goal)
```

```python
#analysis for weekly steps
weeks = np.array_split(steps, 4)
print("Week Total Analysis:")
week_total = [print(week.sum()) for week in weeks]
print("\nWeek Average Analysis:")
week_avg = [print(week.mean()) for week in weeks]
```

```python
#anomaly detection
lower_bound = mean_step - 2 * std_step
upper_bound = mean_step + 2 * std_step
print(lower_bound)
print(upper_bound)

print(round(lower_bound, 2))
print(round(upper_bound, 2))

anomaly = np.where((steps < lower_bound) | (steps > upper_bound)) [0] + 1
print("Anomaly detection (low/high):", anomaly)
```
