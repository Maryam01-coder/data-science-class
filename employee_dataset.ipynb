```python
#start cleaning dataset
#renaming inconsistent columns 
df
```

```python
df.rename(columns={
    "EmpId": "Employee Id",
    "fullName": "Full Name",
    "Dept": "Department",
    "Salary($)": "Salary",
    "JoinDate": "Join Date",
    "Remote?": "Remote"
}, inplace=True)

df
```

```python
#cleaning the inconsistent value of each columns - Full Name
df["Full Name"] = (
    df["Full Name"]
    .str.strip()
    .str.title()
    .str.replace(r"[\s-]+", " ", regex=True)
    .tolist()
)
df["Full Name"]
```

```python
#cleaning inconsistent values in Department
df.replace({
    "hr": "HR",
    "Mktng": "Marketing",
    "marketing": "Marketing",
}, inplace=True)
```

```python
#ceaning inconsistent value in age
df["Age"].to_list()
```

```python
def clean_age(value):
    if value == "thirty-five":
        return 35
    elif isinstance(value, (int, float)):
        return value
    elif isinstance(value, str) and value.isdigit():
        return int(value)
    else: 
        return np.nan

df["Age"] = df["Age"].apply(clean_age)
df["Age"]
```

```python
df
```

```python
#cleaning inconsistent data in salary
df["Salary"] = (
    df["Salary"]
    .astype(str)
    .replace('[$,]', "", regex=True)
    .replace("Nan", np.nan)
    .astype(float)
)

df["Salary"].to_list()
```

```python
#cleaning date values in Join Date
from dateutil import parser

def clean_date(date_str):
    try:
        return parser.parse(date_str, dayfirst=True)
    except Exception:
        return pd.NaT

df["Join Date"] = df["Join Date"].apply(clean_date)
df["Join Date"]
```

```python
# cleaning the performance column
df["Performance"] = (
    df["Performance"]
    .astype(str)
    .replace("%", "", regex=True)
    .replace("None", np.nan)
    .astype(float)
)

df["Performance"]
```

```python
#cleaning the remote column
df.replace({
    "Y": "Yes",
    "N": "No",
    "yes": "Yes",
    "no": "No",
    "Nan": np.nan,
    "True": "Yes"
}, inplace=True)

df["Remote"]
```

```python
df["Employee Id"] = pd.to_numeric(df["Employee Id"], errors="coerce")
df["Employee Id"]
```

```python
df.info()
```

```python
#handling missing values in columns
df["Full Name"] = df["Full Name"].fillna("Unknown")

df["Department"] = df["Department"].fillna(df["Department"].mode()[0])

df["Age"] = df["Age"].fillna(df["Age"].median())

df["Salary"] = df["Salary"].fillna(df["Salary"].median())

df["Performance"] = df["Performance"].fillna(df["Performance"].median())

df["Remote"] = df["Remote"].fillna(df["Remote"].mode()[0])
```

**DATA VISUALIZATION USING PYTHON LIBRARIES - Matplotlib $ Seaborn**

```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
#visualizing average salary by department
plt.bar(merge_df["Department"], merge_df["Department Avg Salary"])
plt.title("Average Salary by Department")
plt.xlabel("Departmet")
plt.ylabel("Average Salary")
plt.grid(axis='y', linestyle="--", alpha=0.7)
plt.show()
```

```python
#view proporttion of remote vs onsite 
remote_status = ["Remote", "Onsite"]
plt.pie(merge_df["Remote"].value_counts(), labels=remote_status, autopct="%1.1f%%")
plt.title("Distribution for Remote Employee")
plt.show()
```

```python
#visualize stack chart for average salary and department budget
plt.bar(merge_df["Department"], merge_df["Department Budget"], color="lightblue", label="Budget")
plt.bar(merge_df["Department"], merge_df["Department Avg Salary"], color="green", label="Salary")
plt.title("Average Salary and Department Budget by Department")
plt.xlabel("Departmet")
plt.ylabel("Average Salary & Department Budget")
plt.show()
```

```python
#visualize employee by performnace overtime
plt.plot(merge_df["Full Name"], merge_df["Performance"], color="brown", marker="o")
plt.title("Employee by Performance")
plt.xlabel("Employee Name")
plt.ylabel("Performance")
plt.xticks(rotation=90)
plt.grid(axis='y', linestyle="-", alpha=0.7)
plt.show()
```

```python
#visualizing with scatter plot - checking correlation between two columns
plt.scatter(merge_df["Salary"], merge_df["Performance"], color="blue")
plt.title("Correlation between Salary and Perfromance")
plt.xlabel("Salary")
plt.ylabel("Performance")
plt.grid(linestyle="--", alpha=0.7)
plt.show()
```

```python
#visulaize the correlation matrix for numerical columns - HEAT MAP
num_cols = ["Age", "Salary", "Performance", "Department Budget", "Department Avg Salary"]
corr = merge_df[num_cols].corr()
corr
```

```python
#visualize the relationship between the employee numerical columns - PAIR PLOT
sns.pairplot(merge_df[num_cols], diag_kind="kde")
plt.suptitle("Pair Plot for Numerical Columns", y=1)
plt.show()
```
